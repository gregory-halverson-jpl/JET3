```python
from dateutil import parser
import matplotlib.pyplot as plt
import rasters as rt
from GEOS5FP import GEOS5FPConnection
from JET3 import process_JET_table, load_ECOv002_calval_JET_inputs
```

```python
RH_filename = "~/Library/CloudStorage/OneDrive-JPL/data/ECOv003_L3T_ETAUX_41310_012_11SPS_20251018T192142_01_RH.tif"
```

```python
GEOS5FP = GEOS5FPConnection()
```

```python
timestamp = RH_filename.split("_")[-3]
time_UTC = parser.parse(timestamp)
time_UTC
```

```python
RH_downscaled = rt.Raster.open(RH_filename)
RH_downscaled.cmap = "viridis_r"
RH_downscaled
```

```python
geometry = RH_downscaled.geometry
geometry
```

```python
RH_coarse = GEOS5FP.RH(time_UTC=time_UTC, geometry=geometry)
RH_coarse.cmap = "viridis_r"
RH_coarse
```

```python
%matplotlib inline
fig, axes = plt.subplots(1, 2, figsize=(12, 6))

# Coarse RH on the left
axes[0].imshow(RH_coarse.squeeze(), cmap="viridis_r")
axes[0].set_title("Coarse RH from GEOS5FP")
axes[0].axis('off')

# Downscaled RH on the right
axes[1].imshow(RH_downscaled.squeeze(), cmap="viridis_r")
axes[1].set_title("Downscaled RH")
axes[1].axis('off')

plt.tight_layout()
plt.show()
plt.close()
```

```python
%matplotlib inline

import matplotlib.pyplot as plt
import matplotlib.dates as mdates
from matplotlib.ticker import FuncFormatter

# Helper function to generate ordinal day suffix
def ordinal_date_formatter(x, pos=None):
    dt = mdates.num2date(x)
    day = dt.day
    if 11 <= day <= 13:
        suffix = 'th'
    else:
        suffix = {1: 'st', 2: 'nd', 3: 'rd'}.get(day % 10, 'th')
    return f"{dt.strftime('%b')} {day}{suffix}"

# Plot main trend line and uncertainty ribbon
plt.plot(month_df.time_UTC, month_df.ET_daylight_kg, color='black', label='ET Daylight')
plt.fill_between(
    month_df.time_UTC,
    month_df.ET_daylight_kg - month_df.ET_uncertainty,
    month_df.ET_daylight_kg + month_df.ET_uncertainty,
    color='blue',
    alpha=0.2,
    label='ET Uncertainty'
)

# Plot additional model lines
plt.plot(month_df.time_UTC, month_df.ET_daylight_BESS_kg, color='orange', label='BESS')
plt.plot(month_df.time_UTC, month_df.ET_daylight_STIC_kg, color='green', label='STIC')
plt.plot(month_df.time_UTC, month_df.ET_daylight_PTJPLSM_kg, color='purple', label='PTJPLSM')
plt.plot(month_df.time_UTC, month_df.ET_daylight_PMJPL_kg, color='red', label='PMJPL')

# Apply custom x-axis date format
plt.gca().xaxis.set_major_formatter(FuncFormatter(ordinal_date_formatter))

# Rotate labels -45 degrees and align them to the tick mark
plt.xticks(rotation=-45, ha='left')

plt.ylim(bottom=0)
plt.legend()
```

```python
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
import matplotlib.ticker as ticker

%matplotlib inline

# Helper function to generate ordinal day suffix
def format_ordinal_date(x, pos=None):
    dt = mdates.num2date(x)
    day = dt.day
    suffix = 'th' if 11 <= day <= 13 else {1: 'st', 2: 'nd', 3: 'rd'}.get(day % 10, 'th')
    return f"{dt.strftime('%b')} {day}{suffix}"

# 1. Setup figure and update GridSpec height ratios
fig = plt.figure(figsize=(12, 12))

# height_ratios=[2, 1, 1] allocates double height to the top row and thins out the bottom row
gs = fig.add_gridspec(3, 2, height_ratios=[2, 1, 1])

# --- ROW 0: RH Comparison (Enlarged) ---
ax00 = fig.add_subplot(gs[0, 0])
ax00.imshow(RH_coarse.squeeze(), cmap="viridis_r")
ax00.set_title("Example of Coarse Relative Humidity from GEOS-5 FP")
ax00.axis('off')

ax01 = fig.add_subplot(gs[0, 1])
ax01.imshow(RH_downscaled.squeeze(), cmap="viridis_r")
ax01.set_title("Example of Downscaled Relative Humidity")
ax01.axis('off')

# --- ROW 1: LE and ET Comparison ---
ax10 = fig.add_subplot(gs[1, 0])
ax10.plot(month_df.time_UTC, month_df.LE_Wm2, label="LE_Wm2", color="blue")
ax10.set_title("Example of Instantaneous Latent Heat Flux")
ax10.set_xlabel("Date")
ax10.set_ylabel("LE (W/m²)")

ax11 = fig.add_subplot(gs[1, 1])
ax11.bar(month_df.time_UTC, month_df.ET_daylight_kg, label="ET_daylight_kg", color="blue")
ax11.set_title("Example of Daily Upscaled Evapotranspiration")
ax11.set_xlabel("Date")
ax11.set_ylabel("ET (mm/day)")

# --- ROW 2: Model Comparison (Thinner full-width row) ---
ax_bottom = fig.add_subplot(gs[2, :])

ax_bottom.plot(month_df.time_UTC, month_df.ET_daylight_kg, color='black', label='ET Daylight')
ax_bottom.fill_between(
    month_df.time_UTC,
    month_df.ET_daylight_kg - month_df.ET_uncertainty,
    month_df.ET_daylight_kg + month_df.ET_uncertainty,
    color='blue',
    alpha=0.2,
    label='ET Uncertainty'
)

ax_bottom.plot(month_df.time_UTC, month_df.ET_daylight_BESS_kg, color='orange', label='BESS')
ax_bottom.plot(month_df.time_UTC, month_df.ET_daylight_STIC_kg, color='green', label='STIC')
ax_bottom.plot(month_df.time_UTC, month_df.ET_daylight_PTJPLSM_kg, color='purple', label='PTJPLSM')
ax_bottom.plot(month_df.time_UTC, month_df.ET_daylight_PMJPL_kg, color='red', label='PMJPL')

ax_bottom.set_title("Example of Multi-Model Evapotranspiration Uncertainty")
ax_bottom.set_xlabel("Date")
ax_bottom.set_ylabel("ET (mm/day)")
ax_bottom.set_ylim(bottom=0)
ax_bottom.legend()

# --- Format Date Axes for Rows 1 and 2 ---
date_fmt = ticker.FuncFormatter(format_ordinal_date)
for ax in (ax10, ax11, ax_bottom):
    ax.xaxis.set_major_formatter(date_fmt)
    plt.setp(ax.get_xticklabels(), rotation=-45, ha='left')

plt.tight_layout()
plt.show()
plt.close()
```
